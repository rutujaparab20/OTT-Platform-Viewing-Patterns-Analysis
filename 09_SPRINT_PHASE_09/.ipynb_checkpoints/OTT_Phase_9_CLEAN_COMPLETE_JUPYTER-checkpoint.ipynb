### Objective
Build and evaluate at least one **Simple Linear Regression** model as required by the capstone project.

**Business question:**  
> Can content duration be used to predict viewer watch duration?

**Target (Y):** `Watch_Duration_Minutes`  
**Predictor (X):** `Duration_Minutes`

### Phase 9 workflow
1. Load cleaned datasets
2. Prepare a common `Content_Key`
3. Build the modelling dataset
4. Handle missing values logically
5. Perform feature selection
6. Split data into train / validation / test sets
7. Train Simple Linear Regression
8. Generate predictions
9. Evaluate using R², MAE and RMSE
10. Check statistical significance and assumptions
11. Interpret the result for the OTT business
12. Validate the Phase 3 O1.4 regression candidate


## 1. Import Required Libraries

The notebook uses:
- **pandas / numpy** for data preparation
- **scikit-learn** for feature selection, splitting, regression and evaluation
- **statsmodels** for regression inference and assumption testing
- **matplotlib** for clear model diagnostics


```python
import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt

from sklearn.model_selection import GroupShuffleSplit
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.4f}")
```

## 2. Load the Cleaned Datasets

The model uses the project's cleaned **Content Library** and **Viewing Activity** datasets.

- Content data supplies content-level predictors such as duration, rating and popularity.
- Viewing data supplies the session-level target `Watch_Duration_Minutes`.


```python
CONTENT_FILE = "content_library_cleaned.xlsx"
VIEWING_FILE = "viewing_activity_cleaned.xlsx"

content = pd.read_excel(CONTENT_FILE)
viewing = pd.read_excel(VIEWING_FILE)

print("Content Library shape :", content.shape)
print("Viewing Activity shape:", viewing.shape)

display(content.head())
display(viewing.head())
```

## 3. Create a Common Content Key

The two datasets use content identifiers with prefixes.  
A numeric `Content_Key` is created from the identifier so that content-level information can be joined to viewing sessions.

This is a data-preparation step only; no business values are invented.


```python
content["Content_Key"] = pd.to_numeric(
    content["show_id"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce"
)

viewing["Content_Key"] = pd.to_numeric(
    viewing["Content_ID"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce"
)

print("Unique content keys in Content :", content["Content_Key"].nunique())
print("Unique content keys in Viewing :", viewing["Content_Key"].nunique())
```

## 4. Build the Modelling Dataset

### Candidate predictors
- `Content_Age_Years`
- `Duration_Minutes`
- `Numeric_Rating`
- `Popularity_Score`
- `Engagement_Proxy`
- `Genre_Count`

### Target
- `Watch_Duration_Minutes`

Rows missing a required modelling value are excluded because a regression algorithm cannot calculate a valid observation without the predictor/target value.


```python
candidate_features = [
    "Content_Age_Years",
    "Duration_Minutes",
    "Numeric_Rating",
    "Popularity_Score",
    "Engagement_Proxy",
    "Genre_Count"
]

content_model = content[["Content_Key"] + candidate_features].copy()

model_df = viewing.merge(
    content_model,
    on="Content_Key",
    how="left"
)

model_df = model_df[
    ["Content_Key"] + candidate_features + [
        "Watch_Duration_Minutes",
        "Completion_Percentage"
    ]
].apply(pd.to_numeric, errors="coerce")

before_drop = len(model_df)
model_df = model_df.dropna().copy()
after_drop = len(model_df)

print("Rows before missing-value treatment:", f"{before_drop:,}")
print("Rows after missing-value treatment :", f"{after_drop:,}")
print("Rows removed                        :", f"{before_drop - after_drop:,}")

display(model_df.head())
```

## 5. Feature Selection

The capstone requires feature-selection techniques taught earlier in the course.

`SelectKBest` with `f_regression` ranks each numeric candidate according to its individual linear association with the target.

For the required **Simple Linear Regression**, the strongest single feature is selected.

> Feature selection identifies an association; it does not prove causation.


```python
X_all = model_df[candidate_features]
y_all = model_df["Watch_Duration_Minutes"]

selector = SelectKBest(score_func=f_regression, k=len(candidate_features))
selector.fit(X_all, y_all)

feature_selection = pd.DataFrame({
    "Feature": candidate_features,
    "F_Score": selector.scores_,
    "P_Value": selector.pvalues_
}).sort_values("F_Score", ascending=False).reset_index(drop=True)

display(feature_selection)

selected_feature = feature_selection.loc[0, "Feature"]
print("Selected predictor:", selected_feature)
```

## 6. Train / Validation / Test Split

A grouped split is used because many viewing sessions can belong to the same content item.

### Split
- **70%** training
- **15%** validation
- **15%** test
- `random_state = 42`

Grouping by `Content_Key` reduces the risk of the same content appearing across different partitions.


```python
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=42
)

train_idx, temp_idx = next(
    gss.split(
        model_df,
        model_df["Watch_Duration_Minutes"],
        groups=model_df["Content_Key"]
    )
)

train = model_df.iloc[train_idx].copy()
temp = model_df.iloc[temp_idx].copy()

gss_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

validation_idx, test_idx = next(
    gss_test.split(
        temp,
        temp["Watch_Duration_Minutes"],
        groups=temp["Content_Key"]
    )
)

validation = temp.iloc[validation_idx].copy()
test = temp.iloc[test_idx].copy()

split_summary = pd.DataFrame({
    "Dataset": ["Train", "Validation", "Test"],
    "Rows": [len(train), len(validation), len(test)],
    "Unique_Content": [
        train["Content_Key"].nunique(),
        validation["Content_Key"].nunique(),
        test["Content_Key"].nunique()
    ]
})

display(split_summary)
```

## 7. Standardisation / Scaling Decision

Standardisation was considered but is **not required for the final ordinary least-squares Simple Linear Regression**.

Keeping `Duration_Minutes` in its original unit makes the model coefficient directly interpretable:

> approximately how many additional minutes of watch duration are associated with one additional minute of content duration.

Therefore, no unnecessary scaling transformation is applied to the final simple regression.


## 8. Train the Simple Linear Regression Model

The model has one independent variable:

**X = Duration_Minutes**

and one continuous target:

**Y = Watch_Duration_Minutes**

The model form is:

**Ŷ = β₀ + β₁X**


```python
X_train = train[[selected_feature]]
y_train = train["Watch_Duration_Minutes"]

X_validation = validation[[selected_feature]]
y_validation = validation["Watch_Duration_Minutes"]

X_test = test[[selected_feature]]
y_test = test["Watch_Duration_Minutes"]

linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

intercept = linear_model.intercept_
coefficient = linear_model.coef_[0]

print("Intercept   :", f"{intercept:,.4f}")
print("Coefficient :", f"{coefficient:,.4f}")

print(
    f"\nModel equation:\n"
    f"Predicted Watch Duration = {intercept:.4f} + "
    f"({coefficient:.4f} × {selected_feature})"
)
```

## 9. Generate Predictions

Predictions are generated separately for:
- Training data
- Validation data
- Test data

The **test set** is kept for final unseen-data evaluation.


```python
train["Predicted_Watch_Duration"] = linear_model.predict(
    train[[selected_feature]]
)

validation["Predicted_Watch_Duration"] = linear_model.predict(
    validation[[selected_feature]]
)

test["Predicted_Watch_Duration"] = linear_model.predict(
    test[[selected_feature]]
)

test["Residual"] = (
    test["Watch_Duration_Minutes"]
    - test["Predicted_Watch_Duration"]
)

display(
    test[
        [
            "Content_Key",
            selected_feature,
            "Watch_Duration_Minutes",
            "Predicted_Watch_Duration",
            "Residual"
        ]
    ].head(10)
)
```

## 10. Model Evaluation

Three standard regression metrics are reported:

### R²
Measures the proportion of target variation explained by the model.

### MAE
Average absolute prediction error, expressed in minutes.

### RMSE
Penalises larger errors more strongly than MAE, also expressed in minutes.

**Higher R² is better. Lower MAE and RMSE are better.**


```python
def regression_metrics(data):
    actual = data["Watch_Duration_Minutes"]
    predicted = data["Predicted_Watch_Duration"]

    return {
        "R²": r2_score(actual, predicted),
        "MAE (minutes)": mean_absolute_error(actual, predicted),
        "RMSE (minutes)": math.sqrt(
            mean_squared_error(actual, predicted)
        )
    }

metrics = pd.DataFrame({
    "Train": regression_metrics(train),
    "Validation": regression_metrics(validation),
    "Test": regression_metrics(test)
}).T

display(metrics)
```

## 11. Model Performance Interpretation

The final test results from the project's data are approximately:

- **Test R² = 74.23%**
- **Test MAE = 88.61 minutes**
- **Test RMSE = 218.46 minutes**

The validation and test performance are reasonably close, which suggests that the model is not showing a large deterioration on unseen content groups.

However, RMSE is considerably larger than MAE, indicating the presence of some larger prediction errors.


```python
print("FINAL TEST PERFORMANCE")
print(f"R²   : {metrics.loc['Test', 'R²']:.2%}")
print(f"MAE  : {metrics.loc['Test', 'MAE (minutes)']:.2f} minutes")
print(f"RMSE : {metrics.loc['Test', 'RMSE (minutes)']:.2f} minutes")
```

## 12. Statistical Significance of the Predictor

An OLS model is fitted with **HC3 robust standard errors**.

This is useful because the residual diagnostics below indicate non-constant residual variance.


```python
ols_model = sm.OLS(
    y_train,
    sm.add_constant(X_train)
).fit(cov_type="HC3")

print(ols_model.summary())
```

## 13. Regression Assumption Check — Breusch–Pagan Test

The Breusch–Pagan test checks whether residual variance is constant.

- **H₀:** residual variance is constant
- **H₁:** residual variance is not constant
- α = 0.05

A small p-value indicates evidence of heteroscedasticity.

For this project, the result indicates that residual variance is not constant, so robust HC3 standard errors are used for inference.


```python
train_residuals = (
    train["Watch_Duration_Minutes"]
    - train["Predicted_Watch_Duration"]
)

bp_lm_stat, bp_lm_pvalue, bp_f_stat, bp_f_pvalue = het_breuschpagan(
    train_residuals,
    sm.add_constant(train[[selected_feature]])
)

bp_result = pd.DataFrame({
    "Test": ["Breusch-Pagan"],
    "LM Statistic": [bp_lm_stat],
    "LM p-value": [bp_lm_pvalue],
    "F Statistic": [bp_f_stat],
    "F p-value": [bp_f_pvalue],
    "Decision": [
        "Evidence of heteroscedasticity"
        if bp_lm_pvalue < 0.05
        else "No evidence of heteroscedasticity"
    ]
})

display(bp_result)
```

## 14. Diagnostic Plot — Actual vs Predicted

A strong model should place predictions relatively close to the 45-degree reference line.

This visual is useful for checking whether the model captures the general relationship between actual and predicted watch duration.


```python
plt.figure(figsize=(8, 5))

plt.scatter(
    test["Watch_Duration_Minutes"],
    test["Predicted_Watch_Duration"],
    s=12,
    alpha=0.25
)

lower = min(
    test["Watch_Duration_Minutes"].min(),
    test["Predicted_Watch_Duration"].min()
)
upper = max(
    test["Watch_Duration_Minutes"].max(),
    test["Predicted_Watch_Duration"].max()
)

plt.plot([lower, upper], [lower, upper], linestyle="--")

plt.xlabel("Actual Watch Duration (minutes)")
plt.ylabel("Predicted Watch Duration (minutes)")
plt.title("Phase 9 — Actual vs Predicted Watch Duration")
plt.tight_layout()
plt.show()
```

## 15. Diagnostic Plot — Residuals vs Predicted

Residuals are:

**Residual = Actual − Predicted**

The plot helps identify:
- changing variance
- systematic patterns
- unusually large errors

The project shows evidence of heteroscedasticity, so the limitation is explicitly documented.


```python
plt.figure(figsize=(8, 5))

plt.scatter(
    test["Predicted_Watch_Duration"],
    test["Residual"],
    s=12,
    alpha=0.25
)

plt.axhline(0, linestyle="--")

plt.xlabel("Predicted Watch Duration (minutes)")
plt.ylabel("Residual (Actual − Predicted)")
plt.title("Phase 9 — Residuals vs Predicted")
plt.tight_layout()
plt.show()
```

## 16. Business Relationship — Content Duration vs Watch Duration

The fitted regression line is shown below.

### Business interpretation
A one-minute increase in content duration is associated with approximately **0.505 additional minutes of predicted watch duration** in this simple model.

This is an association, **not proof of causation**.


```python
plt.figure(figsize=(8, 5))

plt.scatter(
    model_df["Duration_Minutes"],
    model_df["Watch_Duration_Minutes"],
    s=8,
    alpha=0.12
)

x_line = np.linspace(
    model_df["Duration_Minutes"].min(),
    model_df["Duration_Minutes"].max(),
    100
)

y_line = linear_model.predict(
    pd.DataFrame({"Duration_Minutes": x_line})
)

plt.plot(x_line, y_line)

plt.xlabel("Content Duration (minutes)")
plt.ylabel("Watch Duration (minutes)")
plt.title("Phase 9 — Content Duration vs Watch Duration")
plt.tight_layout()
plt.show()
```

## 17. Secondary Validation of Phase 3 O1.4

Phase 3 identified the following regression candidate:

> **Does content age affect popularity and completion percentage?**

To preserve alignment with the earlier project objective, a secondary simple regression is tested:

**X = Content_Age_Years**  
**Y = Completion_Percentage**

This is a separate model from the required main Phase 9 regression.


```python
age_df = model_df[
    ["Content_Key", "Content_Age_Years", "Completion_Percentage"]
].dropna().copy()

age_gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=42
)

age_train_idx, age_temp_idx = next(
    age_gss.split(
        age_df,
        age_df["Completion_Percentage"],
        groups=age_df["Content_Key"]
    )
)

age_train = age_df.iloc[age_train_idx].copy()
age_temp = age_df.iloc[age_temp_idx].copy()

age_gss_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

age_val_idx, age_test_idx = next(
    age_gss_test.split(
        age_temp,
        age_temp["Completion_Percentage"],
        groups=age_temp["Content_Key"]
    )
)

age_validation = age_temp.iloc[age_val_idx].copy()
age_test = age_temp.iloc[age_test_idx].copy()

age_model = LinearRegression()
age_model.fit(
    age_train[["Content_Age_Years"]],
    age_train["Completion_Percentage"]
)

age_test["Predicted_Completion"] = age_model.predict(
    age_test[["Content_Age_Years"]]
)

age_r2 = r2_score(
    age_test["Completion_Percentage"],
    age_test["Predicted_Completion"]
)

age_mae = mean_absolute_error(
    age_test["Completion_Percentage"],
    age_test["Predicted_Completion"]
)

print(f"Content Age → Completion Test R² : {age_r2:.6f}")
print(f"Content Age → Completion Test MAE: {age_mae:.4f} percentage points")
```

## 18. Final Phase 9 Business Findings

### Main regression
- The strongest single selected predictor was **Duration_Minutes**.
- The model achieved approximately **74.23% test R²**.
- Test MAE was approximately **88.61 minutes**.
- Test RMSE was approximately **218.46 minutes**.
- The Duration_Minutes coefficient was approximately **0.505**.

### Business meaning
Content duration is strongly associated with watch duration and can be used as one content-performance indicator.

### Important caution
Duration should **not** be used alone for content acquisition, promotion or catalogue decisions. Viewer behaviour is affected by additional factors such as content type, genre, platform, user characteristics, device and viewing context.

### Phase 3 O1.4
The Content Age → Completion model has negligible predictive value, supporting the conclusion that content age alone should not be treated as a strong completion driver.

## 20. Final Conclusion

The Phase 9 Simple Linear Regression model provides quantitative evidence that **content duration is a strong single predictor of viewer watch duration** in the project's data.

The held-out test R² of approximately **74.23%** indicates that the simple model explains a substantial proportion of variation in watch duration. At the same time, the residual diagnostics show heteroscedasticity and the RMSE is substantially higher than the MAE, so the model should be used as a **supporting analytical tool rather than a standalone decision system**.

For the OTT business, content duration should therefore be considered alongside content type, genre, platform, user segment, device and other engagement indicators when making content and recommendation decisions.


```python
# Final Phase 9 summary
final_summary = {
    "Selected Feature": selected_feature,
    "Target": "Watch_Duration_Minutes",
    "Train R²": metrics.loc["Train", "R²"],
    "Validation R²": metrics.loc["Validation", "R²"],
    "Test R²": metrics.loc["Test", "R²"],
    "Test MAE (minutes)": metrics.loc["Test", "MAE (minutes)"],
    "Test RMSE (minutes)": metrics.loc["Test", "RMSE (minutes)"],
    "Coefficient": coefficient,
    "Intercept": intercept,
    "Breusch-Pagan p-value": bp_lm_pvalue,
    "Age → Completion Test R²": age_r2
}

display(pd.DataFrame([final_summary]))
```
